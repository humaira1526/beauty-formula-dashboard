## 4. Visualization techniques

1. **Horizontal bar chart — most frequent ingredients.** It ranks ingredients by the number of products in which they appear. It is useful for comparing a small set of common ingredients, but not for showing relationships or long-tail distributions.
2. **Histogram — ingredient-list length.** It shows how product ingredient counts are distributed. It is useful for shape and spread, but binning can hide individual values and it does not explain which ingredients are present.
3. **Sunburst — category to ingredient function.** Nested sectors show how function labels break down within product categories. It is useful for hierarchical composition, but becomes difficult to read with too many small segments.
4. **Sankey diagram — category-to-function pathways (new to SIADS 521).** Link width represents the number of product–ingredient records connecting a category to a function. Unlike a standard bar chart, it emphasizes the flow/connection structure between two sets of categories. It can become cluttered when too many nodes or links are included.

All charts use the same category filter. This allows a viewer to compare how the ingredient ranking, list-length distribution, functional hierarchy, and category/function pathways change together.
